# Data

Code to load and process the data. 

There are 14 different datasets containing data from Feb-May (inclusive) for the years 2012-2025.

```python
#Load data from different spreadsheets
data = [[] for x in range(14)]

years = range(12,26)
for year in years:
    fileName = 'powerData' + str(year) + '.csv'
    yearData = pd.read_csv(os.path.join(os.getcwd(), 'datasets', fileName))
    
    yearData = yearData.drop('id', axis=1) #Ignore ID column
    
    yearData[' demand'] = yearData[' demand'] / 1000 #Convert to GW
    
    data[year-12] = yearData #Add to list of years' data
```

# Exploratory Data Analysis

Exploratory Data Analysis of your chosen dataset, which should demonstrate that you have understood the structure of the data, identified anomalous or missing data, and used visualisation and descriptive statistics to explore possible relationships in the data. 

```python
#Inspect head of one of the datasets
data[0].head().to_latex("tables/grid-table.tex", index=False, column_format="lr")
```

Only days as a whole will be considered so all five-minute intervals for that day must be summed into a single data point.

```python
#Find daily totals
dailyData = [pd.DataFrame(columns=['Date', 'Demand']) for _ in range(14)] #dataframe of daily totals for each year

dates = [] #all dates in dataset
dates = dates + ['02-0'+str(n) for n in range(1,10)] + ['02-'+str(n) for n in range(10,29)] #Feb
dates = dates + ['03-0'+str(n) for n in range(1,10)] + ['03-'+str(n) for n in range(10,32)] #Mar
dates = dates + ['04-0'+str(n) for n in range(1,10)] + ['04-'+str(n) for n in range(10,31)] #Apr
dates = dates + ['05-0'+str(n) for n in range(1,10)] + ['05-'+str(n) for n in range(10,32)] #May

for year in range(14):
    for date in dates:
        dateTotal = data[year][data[year][' timestamp'].str[6:11] == date][' demand'].sum()
        dailyData[year].loc[(len(dailyData[year]))] = [date, dateTotal] #Append to dataframe
```

There are many missing intervals for the dates 18-20 March 2025. Since these daily totals are so close to the introduction of BST, they cannot be ignored. The previous years' totals for each date is averaged as that should give a good estimation of what the missing data should be.

```python
#Replace missing totals from 18-20 March 2025 with averages on that day from previous years
for date in ['03-18', '03-19', '03-20']:
    sum = 0
    for year in range(13):
        sum += dailyData[year].loc[dailyData[year]['Date']==date].iloc[0]['Demand'] #Sum previous years' data
    dailyData[13].loc[dailyData[13]['Date']==date] = [date, sum/13] #Update with average
```

```python
#Export head of final daily total data
output = dailyData[0].head().copy()
output.rename(columns={'Demand': 'Demand (GW)'}, inplace=True)
output.to_latex("tables/daily-table.tex", index=False, column_format="lcccr")
```

Power demand was found to vary dramatically between weekend days and weekdays (weekends have lower consumption). To improve the regression model, the data should contain a Boolean factor for this information. `from datetime import date as dateFinder` aids with finding weekend status for each day.

```python
def isWeekend(year, date): #True for Sat/Sun; False otherwise
    #assumes year is 2012-based and date is 'mm-dd'
    dayOfWeek = dateFinder(year+2012, int(date[:2]), int(date[3:5])).weekday()
    return (dayOfWeek == 5 or dayOfWeek == 6)

#Add column to signify weekend inclusion.
for year in range(14):
    weekend = [isWeekend(year, date) for date in dailyData[year]['Date']]
    dailyData[year].insert(2, 'Weekend', weekend) #add column to dataframe
```

The regression model must know whether a date occurs before or after the introduction of BST. A column is also added for an index rebased around zero at the change date, which is mostly for visualisation purposes.

```python
#Add column to rebase index around the change date (- before, + after)
#Add column to signify whether date is before (0) or after (1) the intro of BST.

BSTDates = [25, 31, 30, 29, 27, 26, 25, 31, 29, 28, 27, 26, 31, 30] #Day in March of the given year where BST introduced

for year in range(14):
    changeDateIndex = dailyData[year].loc[dailyData[year]['Date'] == '03-' + str(BSTDates[year])].index[0]
    #Index rebasing
    index = [(i - changeDateIndex) for i in range(len(dailyData[year]))] 
    dailyData[year].insert(3, 'Index', index)
    #BST Boolean
    BST = [False]*changeDateIndex + [True]*(len(dailyData[year])-changeDateIndex)  
    dailyData[year].insert(3, 'BST', BST)
```

```python
#Export head of final daily total data
output = dailyData[0].head().copy()
output.rename(columns={'Demand': 'Demand (GW)'}, inplace=True)
output.to_latex("tables/final-daily-table.tex", index=False, column_format="lcccr")
```

# Data communication and analysis

The visualisation(s), tables and any statistical analyses or application of ML to the data, as  presented in your report. 

```python
#Show visualisation of split between BST and non-BST datapoints and splot between weekend and weekday 
#data points to justify terms in regression model.
#NB: colours used are colour-blind friendly.
for year in range(14):

    dataSubset = dailyData[year][(~dailyData[year]['BST']) & (~dailyData[year]['Weekend'])] #Non-BST Weekdays
    plt.scatter('Index', 'Demand', c='#009E73', marker='o', data=dataSubset, label='Non-BST Weekdays') 
    
    dataSubset = dailyData[year][(~dailyData[year]['BST']) & (dailyData[year]['Weekend'])] #Non-BST Weekends
    plt.scatter('Index', 'Demand', c='#009E73', marker='x', data=dataSubset, label='Non-BST Weekends') 
    
    dataSubset = dailyData[year][(dailyData[year]['BST']) & (~dailyData[year]['Weekend'])] #BST Weekdays
    plt.scatter('Index', 'Demand', c='#56B4E9', marker='o', data=dataSubset, label='BST Weekdays') 
    
    dataSubset = dailyData[year][(dailyData[year]['BST']) & (dailyData[year]['Weekend'])] #BST Weekends
    plt.scatter('Index', 'Demand', c='#56B4E9', marker='x', data=dataSubset, label='BST Weekends') 
    

    #Captions
    plt.title(f"Power Demand Around Clock Change {2012+year}")
    plt.xlabel("Date Rebased (0 = BST Introduction Date)")
    plt.ylabel("Daily Power Demand (GW)")
    plt.legend(loc='best')

    plt.savefig(f'visualisations/year{year}Splits.pdf')
    plt.close()
```

A multiple regression model is performed to predict demand based on the date, weekend status, and whether it is before or during BST. An intercation term is added between BST and Index as the general slope may change after BST is introduced.

```python
#Fit linear regression and collect coefficients
result = [0 for _ in range(14)] #Fitted model for each year
parameterRows = [0 for _ in range(14)]

for year in range(14):
    model = smf.ols('Demand ~ Index + BST + (BST * Index) + Weekend', dailyData[year])
    result[year] = model.fit()
    parameterRows[year] = result[year].params #Collect coefficients
    parameterRows[year].loc['R^2'] = result[year].rsquared
    
#Create dataframe
params = pd.DataFrame(parameterRows)
params.loc['Year'] = [year for year in range(2012,2026)]
```

```python
#Export regression parameters
output = params.head().copy()
output.rename(columns={'Intercept': 'Intercept (GW)', 'BST[T.True]': 'BST (GW)', 'Weekend[T.True]': 'Weekend (GW)'}, inplace=True)
output.rename(columns={'Index': 'Index (GW/Day)', 'BST[T.True]:Index': 'BST Index (GW/Day)'}, inplace=True)
output.to_latex("tables/regression-table.tex", index=False, column_format="lcccccr", float_format="%.2f", escape=True)
```

The key statistic of `Drop` is calculated: the reduction attributable to the introduction of BST compared with what is predicted to happen if the clocks did not change. 

```python
#Add percentage drop column
params['Drop'] = (1-(params['Intercept']+params['BST[T.True]']) / params['Intercept'])*100
```

```python
#Export prediction data
output = {'Year': [2012+i for i in range(14)],
          'Factual Estimate (GW)': params['Intercept']+params['BST[T.True]'],
          'Counterfactual Estimate (GW)': params['Intercept'],
          'Drop (%)': params['Drop']}
outputDF = pd.DataFrame(output)
outputDF.to_latex("tables/predictions-table.tex", index=False, column_format="lccr", float_format="%.2f", escape=True)
```

A visualtion shows the counterfactual green line of predictions had BST not been introduced and the factual blue line of the predictions accounting for BST as a factor. The red line shows the intial difference between these predictions.

```python
#Produce visualisations showing daily totals before and after clock change
#NB: colours used are colour-blind friendly.
for year in range(14):
    plt.scatter('Index', 'Demand', c='#000000', data=dailyData[year]) #Plot all daily total points

    #Counterfactual as if BST was not introduced
    beforeXs = dailyData[year]['Index'].tolist() #Days
    predBefore = pd.DataFrame({"Index": beforeXs, "BST": 0, "Weekend": 0}) #Prediction dataframe with BST fixed to false
    predBefore["Predicted"] = result[year].predict(predBefore) #Demand predictions
    plt.plot(beforeXs, predBefore['Predicted'], color='#009E73', label='Counterfactual lines') #Green line

    #Counterfactual as if BST was not introduced -- Weekend
    predBefore = pd.DataFrame({"Index": beforeXs, "BST": 0, "Weekend": 1}) #Prediction dataframe with BST fixed to false
    predBefore["Predicted"] = result[year].predict(predBefore) #Demand predictions
    plt.plot(beforeXs, predBefore['Predicted'], color='#009E73') #Green line

    #Demand after BST was introduced
    afterXs = dailyData[year][dailyData[year]['Index']>=0]['Index'].tolist() #Days
    predAfter = pd.DataFrame({"Index": afterXs, "BST": 1, "Weekend": 0}) #Prediction dataframe
    predAfter["Predicted"] = result[year].predict(predAfter) #Demand predictions
    plt.plot(afterXs, predAfter['Predicted'], color='#56B4E9', label='Factual lines') #Blue line

    #Demand after BST was introduced -- Weekend
    predAfter = pd.DataFrame({"Index": afterXs, "BST": 1, "Weekend": 1}) #Prediction dataframe
    predAfter["Predicted"] = result[year].predict(predAfter) #Demand predictions
    plt.plot(afterXs, predAfter['Predicted'], color='#56B4E9') #Blue line

    #Initial difference between lines at day 0
    topY = params.iloc[year]['Intercept'] #Counterfactual prediction of day 0 demand.
    bottomY = params.iloc[year]['Intercept']+params.iloc[year]['BST[T.True]'] #Factual prediction
    plt.vlines(x=0, ymin=bottomY, ymax=topY, colors='#D55E00', linestyles='dotted', label='Initial difference') #Orange line

    #Captions
    plt.title(f"Power Demand Regressions Around Clock Change {2012+year}")
    plt.xlabel("Date Rebased (0 = BST Introduction Date)")
    plt.ylabel("Daily Power Demand (GW)")
    plt.legend(loc='best')

    plt.savefig(f'visualisations/year{year}.pdf')
    plt.close()
```

```python
#Calculate confidence intervals
def CIs(title, values):
    t_critical = 2.1604  # for df=13, 95% CI
    
    mean = statistics.mean(values)
    s = statistics.stdev(values)  #Sample standard deviation
    SEM = s / math.sqrt(len(values))
    
    CI = [mean - t_critical * SEM, mean + t_critical * SEM]
    
    print(f"Mean {title} is {mean:.6f} with the 95% CI between {CI[0]:.6f} and {CI[1]:.6f}")

CIs('BST drop', params['Drop'])
```
