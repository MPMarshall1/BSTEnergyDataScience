# Data

Code to load and process the data. 

```python
#Load data from different spreadsheets
data = [[] for x in range(14)]

years = range(12,26)
for year in years:
    fileName = 'powerData' + str(year) + '.csv'
    yearData = pd.read_csv(os.path.join(os.getcwd(), 'datasets', fileName))
    yearData = yearData.drop('id', axis=1)
    yearData[' demand'] = yearData[' demand'] / 1000 #convert to GW
    data[year-12] = yearData
```

# Exploratory Data Analysis

Exploratory Data Analysis of your chosen dataset, which should demonstrate that you have understood the structure of the data, identified anomalous or missing data, and used visualisation and descriptive statistics to explore possible relationships in the data. 

```python
#Inspect head of one of the datasets
data[0].head().to_latex("tables/grid-table.tex", index=False, column_format="lr")
```

```python
#Find daily totals
dailyTotals = [[] for x in range(14)] #a list of each day's total for each year

dates = [] #all dates in dataset
dates = dates + ['02-0'+str(n) for n in range(1,10)] + ['02-'+str(n) for n in range(10,29)] 
dates = dates + ['03-0'+str(n) for n in range(1,10)] + ['03-'+str(n) for n in range(10,32)] 
dates = dates + ['04-0'+str(n) for n in range(1,10)] + ['04-'+str(n) for n in range(10,31)]
dates = dates + ['05-0'+str(n) for n in range(1,10)] + ['05-'+str(n) for n in range(10,32)]

for year in range(14):
    for date in dates:
        dateTotal = data[year][data[year][' timestamp'].str[6:11] == date][' demand'].sum()
        dailyTotals[year].append([date, dateTotal])
```

```python
#Replace missing totals from 18-20 March 2025 with averages on that day from previous years
for day in range(45, 48):
    sum = 0
    for year in range(13):
        sum += dailyTotals[year][day][1]
    dailyTotals[13][day][1] = sum / 13
```

```python
#Export head of daily total data
dateDemand = {'Date': [dailyTotals[0][i][0] for i in range(5)], 
              'Total Daily Demand (W)': [dailyTotals[0][i][1] for i in range(5)]}
dateDemandDF = pd.DataFrame(dateDemand)
dateDemandDF.to_latex("tables/daily-table.tex", index=False, column_format="lr")
```

The data must now be split into sections before and after the clock change date to create different models to compare.

```python
BSTDates = [25, 31, 30, 29, 27, 26, 25, 31, 29, 28, 27, 26, 31, 30] #Day in March of the given year where BST introduced

before = [[] for x in range(14)]
after = [[] for x in range(14)]

for year in range(14):

    changeDate = '03-'+str(BSTDates[year]) #date of clock change
    beforeList = [entry for entry in dailyTotals[year] if entry[0] <= changeDate] #split data
    afterList = [entry for entry in dailyTotals[year] if entry[0] >= changeDate]

    beforeDF = pd.DataFrame(beforeList, columns=['Date', 'Consumption']) #create year-long dataframe
    beforeDF.insert(0, 'Index', range(len(beforeDF)))
    
    afterDF = pd.DataFrame(afterList, columns=['Date', 'Consumption'])
    afterDF.insert(0, 'Index', range(len(afterDF)))

    before[year] = beforeDF #adds to lists of dataframes
    after[year] = afterDF
```

```python
#Add column to signify weekend inclusion.

def isWeekend(year, day): 
    #assumes year is 2012-based and date is 'mm-dd'
    dayOfWeek = dateFinder(year+2012, int(day[:2]), int(day[3:5])).weekday()
    return (dayOfWeek == 5 or dayOfWeek == 6)

for year in range(14):
    weekend = [isWeekend(year, day) for day in before[year]['Date']]
    before[year].insert(3, 'Weekend', weekend)

    weekend = [isWeekend(year, day) for day in after[year]['Date']]
    after[year].insert(3, 'Weekend', weekend)
```

```python
#Make before and after dataframes the same size for fair comparison
#In all years the before portion is smaller so I do not add functionality to rectify this case, save an error message.
for year in range(14):
    afterLength = len(after[year])
    beforeLength = len(before[year])
    
    if (afterLength > beforeLength):
        after[year] = after[year][:beforeLength]
    elif (beforeLength > afterLength):
        print('Error: after list is shorter than before list.')
        
```

```python
#Rebase index such that days are numbered around the clock chnage day at 0 with negative indices before and positive after.
for year in range(14):
    length = len(before[year])
    indexRebased = [i for i in range(length)]
    reverse = [-i for i in indexRebased][::-1]
    
    before[year].insert(4, 'IndexRebased', reverse, True)
    after[year].insert(4, 'IndexRebased', indexRebased, True)
```

# Data communication and analysis

The visualisation(s), tables and any statistical analyses or application of ML to the data, as  presented in your report. 

```python
#Fit linear regression and collect coefficients
beforeIntercept = [0 for _ in range(14)]
afterIntercept = [0 for _ in range(14)]
beforeWeekend = [0 for _ in range(14)]
afterWeekend = [0 for _ in range(14)]
beforeIndex = [0 for _ in range(14)]
afterIndex = [0 for _ in range(14)]

beforeResults = [0 for _ in range(14)]
afterResults = [0 for _ in range(14)]

for year in range(14):
    model = smf.ols('Consumption ~ Index + Weekend', before[year])
    beforeResults[year] = model.fit()

    beforeIntercept[year] = beforeResults[year].params['Intercept']
    beforeWeekend[year] = beforeResults[year].params['Weekend[T.True]']
    beforeIndex[year] = beforeResults[year].params['Index']
        
    model = smf.ols('Consumption ~ Index + Weekend', after[year])
    afterResults[year] = model.fit()
    
    afterIntercept[year] = afterResults[year].params['Intercept']
    afterWeekend[year] = afterResults[year].params['Weekend[T.True]']
    afterIndex[year] = afterResults[year].params['Index']
```

```python
LROutputsBefore = {'Year': [2012+offset for offset in range(14)], 
            'Before Intercept (GW)': beforeIntercept,
            'Before Weekend Coefficient (GW/Day)': beforeWeekend,
            'Before Index (GW/Day)': beforeIndex}
LROutputsDFBefore = pd.DataFrame(LROutputsBefore)
LROutputsDFBefore.head().to_latex("tables/before-regression-table.tex", index=False, column_format="lccr", float_format="%.2f")
```

```python
LROutputsAfter = {'Year': [2012+offset for offset in range(14)], 
            'After Intercept (GW)': afterIntercept,
            'After Weekend Coefficient (GW/Day)': afterWeekend,
            'After Index (GW/Day)': afterIndex}
LROutputsDFAfter = pd.DataFrame(LROutputsAfter)
LROutputsDFAfter.head().to_latex("tables/after-regression-table.tex", index=False, column_format="lccr", float_format="%.2f")
```

```python
#Produce visualisations showing daily totals before and after clock change
for year in range(14):
    fig, axs = plt.subplots(1, 2, sharex=True)
    length = len(before[year])
    
    axs[0].scatter(before[year]['IndexRebased'], before[year]['Consumption'], c='#000000')
    axs[0].scatter(after[year]['IndexRebased'], after[year]['Consumption'], c='#000000')

    axs[0].axvline(x=0, color='red', linestyle='--', linewidth=2, label='x = 5')


    axs[1].scatter(before[year]['IndexRebased'], before[year]['Consumption'], c='#000000')
    axs[1].scatter(after[year]['IndexRebased'], after[year]['Consumption'], c='#000000')

    #Weekends before
    xs = [i for i in range(-length,0) if ((before[year].iloc[length+i])['Weekend'])]
    ys = [beforeResults[year].predict(before[year][before[year]['IndexRebased']==x+1]) for x in xs]
    axs[1].plot(xs, ys, color='#009E73')

    #Weekdays before
    xs = [i for i in range(-length,0) if (not (before[year].iloc[length+i])['Weekend'])]
    ys = [beforeResults[year].predict(before[year][before[year]['IndexRebased']==x+1]) for x in xs]
    axs[1].plot(xs, ys, color='#009E73')
    
    #Weekends after
    xs = [i for i in range(0, length) if ((after[year].iloc[i])['Weekend'])]
    ys = [afterResults[year].predict(after[year][after[year]['IndexRebased']==x]) for x in xs]
    axs[1].plot(xs, ys, color='#56B4E9')

    #Weekdays after
    xs = [i for i in range(0, length) if (not (after[year].iloc[i])['Weekend'])]
    ys = [afterResults[year].predict(after[year][after[year]['IndexRebased']==x]) for x in xs]
    axs[1].plot(xs, ys, color='#56B4E9')

    fig.suptitle(f"Power Demand Regressions Around Clock Change {2012+year}")
    fig.supxlabel("Date")
    fig.supylabel("Daily Power Demand (GW)")

    for ax in axs:
        ax.tick_params(axis='x', rotation=45)
        ax.xaxis.set_major_locator(plt.MaxNLocator(5)) 

    fig.savefig(f'visualisations/year{year}.pdf')
    fig.show()
```

```python
#Calculate changes between before and after coefficients
t_critical = 2.1604

dropFactor = [0 for _ in range(14)] #after-estimate as percentage of before-estimate.

for year in range(14):
    date = '03-'+str(BSTDates[year])
    beforePrediction = (beforeResults[year].predict(before[year][before[year]['Date']==date])).iloc[0]
    afterPrediction = (afterResults[year].predict(after[year][after[year]['Date']==date])).iloc[0]
    dropFactor[year] = 1 - (afterPrediction / beforePrediction)

mean = statistics.mean(dropFactor)
SEM = np.var(dropFactor) / math.sqrt(len(dropFactor))
CI = [mean - t_critical*SEM, mean + t_critical*SEM]

print(f'Mean drop factor is {mean} with the 95% confidence placing it between {CI[0]} and {CI[1]}')
```

```python
estimates = pd.DataFrame([], columns=['Year', 'Before Prediction (GW)', 'After Prediction (GW)', 'Drop%'])
for year in range(14):
    date = '03-'+str(BSTDates[year])
    beforePrediction = (beforeResults[year].predict(before[year][before[year]['Date']==date])).iloc[0]
    afterPrediction = (afterResults[year].predict(after[year][after[year]['Date']==date])).iloc[0]
    dropFactor[year] = (1 - (afterPrediction / beforePrediction)) * 100

    estimates.loc[len(estimates.index)] = [str(year+2012), beforePrediction, afterPrediction, dropFactor[year]*100]
estimates.to_latex("tables/predictions-table.tex", index=False, column_format="lccr", float_format="%.3f")
```
